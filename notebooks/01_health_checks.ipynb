```python
# let's check for disguised null values
categorical_columns = df.select_dtypes(include = ["str"]).columns.to_list()
for column in categorical_columns:
    disguised_nulls = df[column].str.lower().isin(['n/a','nan','none','null','unknown','unspecified','undefined','missing'])
    count = disguised_nulls.sum()
    print(f"{column}: {count}")
```

```python
# No disguised null values
```
